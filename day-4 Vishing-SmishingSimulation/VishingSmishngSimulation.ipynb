```python
import random



def generate_vishing_script(target_company, attacker_role, pretext, scenario):
    """
    Generates a vishing awareness-training script.
    The script uses randomized wording so that each generated script
    can be slightly different.
    """

    # Different openers for different situations
    if scenario == "IT Support Pretext":

        openers = [
            f"Hi, this is Alex from IT Support at {target_company}. "
            f"We detected unusual activity related to your {pretext}.",

            f"Hello, I'm Alex from the {attacker_role} team at "
            f"{target_company}. Our system has flagged an issue involving "
            f"your {pretext}.",

            f"Hi, this is Alex from {target_company} IT Support. "
            f"We're contacting you because our security system detected "
            f"unusual activity involving your account."
        ]

        hooks = [
            "To verify the account, could you provide your employee ID and "
            "confirm some account information?",

            "Before we continue, I need to verify that I'm speaking with "
            "the correct employee. Could you confirm your employee ID?",

            "We need to verify your identity before proceeding with the "
            "security check. Could you confirm your employee ID?"
        ]

        urgency_messages = [
            "This needs to be resolved within the next few minutes or "
            "your account may be temporarily locked.",

            "Please resolve this immediately to prevent interruption "
            "of your access.",

            "We recommend dealing with this immediately because the "
            "account may be restricted if the issue remains unresolved."
        ]

        red_flags = [
            "Legitimate IT staff should NEVER ask for your full password.",
            "Unexpected IT calls should be verified through an official "
            "company channel.",
            "Do not trust caller ID as proof of identity because it can "
            "be spoofed.",
            "Never disclose sensitive credentials during an unexpected call."
        ]

    elif scenario == "Bank Fraud Alert Pretext":

        openers = [
            f"Hello, this is Alex from the {attacker_role} at "
            f"{target_company}. We detected a suspicious transaction "
            f"associated with your account.",

            f"Hi, I'm calling from {target_company}'s fraud prevention "
            f"department. Our monitoring system has detected unusual "
            f"activity on your account.",

            f"Hello, this is Alex from {target_company} Fraud Prevention. "
            f"We're contacting you regarding a security alert on your account."
        ]

        hooks = [
            "We need to verify that you are the account holder before we "
            "can explain the alert. Could you confirm some account details?",

            "For security verification, could you confirm information "
            "associated with your account?",

            "Before discussing the transaction, we need to verify your "
            "identity using information associated with the account."
        ]

        urgency_messages = [
            "This issue should be reviewed immediately to prevent "
            "additional suspicious activity.",

            "Please respond quickly because the account may be temporarily "
            "restricted while the alert remains unresolved.",

            "We recommend taking action immediately so that the suspicious "
            "activity can be investigated."
        ]

        red_flags = [
            "Banks should not ask customers to reveal their full password "
            "or authentication codes over an unexpected call.",
            "Do not rely on caller ID to verify a bank representative.",
            "End the call and contact the bank using its official website, "
            "banking app, or card statement.",
            "Never provide sensitive banking information to an unexpected caller."
        ]

    else:
        raise ValueError("Unknown vishing scenario.")

    opener = random.choice(openers)
    hook = random.choice(hooks)
    urgency = random.choice(urgency_messages)

    script = f"""
=== VISHING AWARENESS SCRIPT (TRAINING USE ONLY) ===

Caller Role   : {attacker_role}
Target Org    : {target_company}
Pretext       : {pretext}

[OPENER]
"{opener}"

[HOOK]
"{hook}"

[URGENCY PUSH]
"{urgency}"

[RED FLAGS FOR AWARENESS]
"""

    for flag in red_flags:
        script += f"-> {flag}\n"

    return script




def generate_smishing_script(target_company, pretext, fake_link):
    """
    Generates an SMS phishing awareness-training scenario.
    """

    sms_templates = [
        f"{target_company} Alert: {pretext}. "
        f"Review your account through the official website.",

        f"{target_company} Security Alert: {pretext}. "
        f"Immediate attention may be required.",

        f"{target_company} Notification: {pretext}. "
        f"Please verify the status of your account."
    ]

    sms_text = random.choice(sms_templates)

    red_flags = [
        "Unexpected SMS messages requesting immediate action are a major "
        "warning sign.",
        "Unfamiliar or suspicious links should not be opened.",
        "Urgency and fear of losing access are common social-engineering tactics.",
        "Do not assume a message is legitimate just because it uses the "
        "organization's name.",
        "Verify the message using the organization's official website or "
        "independently obtained contact information."
    ]

    script = f"""
=== SMISHING AWARENESS SCRIPT (TRAINING USE ONLY) ===

Target Org    : {target_company}
Pretext       : {pretext}

[SMS TEXT]
"{sms_text}"

[SIMULATED TRAINING LINK]
{fake_link}

[RED FLAGS FOR AWARENESS]
"""

    for flag in red_flags:
        script += f"-> {flag}\n"

    return script




TRIGGER_LIBRARY = {

    "authority":
        "The attacker impersonates a trusted authority such as IT, "
        "a bank, or a government organization.",

    "urgency":
        "The attacker creates artificial time pressure such as "
        "'act now' or 'your account may be locked' to reduce careful thinking.",

    "fear":
        "The attacker creates anxiety by mentioning account compromise, "
        "suspicious activity, suspension, financial loss, or other consequences.",

    "scarcity":
        "The attacker suggests that the victim has a limited amount of "
        "time or a limited opportunity to resolve an issue.",

    "liking":
        "The attacker uses a friendly or familiar tone to build rapport "
        "and make the victim less suspicious."
}




def analyze_triggers(scenario_name, triggers_used):

    lines = [
        f"\nScenario: {scenario_name}",
        "Psychological triggers used:"
    ]

    for trigger in triggers_used:

        description = TRIGGER_LIBRARY.get(
            trigger,
            "Unknown psychological trigger."
        )

        lines.append(
            f"  - {trigger.capitalize()}: {description}"
        )

    return "\n".join(lines)




if __name__ == "__main__":

 

    it_scenario = {
        "name": "IT Support Pretext",
        "type": "vishing",
        "company": "Sqrock IT",
        "role": "IT Support",
        "pretext": "Password Reset",
        "triggers": [
            "authority",
            "urgency",
            "fear"
        ]
    }

  

    bank_scenario = {
        "name": "Bank Fraud Alert Pretext",
        "type": "vishing",
        "company": "Meridian National Bank",
        "role": "Fraud Prevention Department",
        "pretext": "Suspicious Transaction Alert",
        "triggers": [
            "authority",
            "fear",
            "urgency"
        ]
    }

  
    government_scenario = {
        "name": "Government Tax Refund Pretext",
        "type": "smishing",
        "company": "National Revenue Service",
        "pretext": "You may be eligible for a tax refund",
        "link": "https://tax-refund.example/verify",
        "triggers": [
            "authority",
            "urgency",
            "fear"
        ]
    }


   

    print("#" * 70)
    print("# 1. IT SUPPORT VISHING SCENARIO")
    print("#" * 70)

    print(
        generate_vishing_script(
            it_scenario["company"],
            it_scenario["role"],
            it_scenario["pretext"],
            it_scenario["name"]
        )
    )

    print(
        analyze_triggers(
            it_scenario["name"],
            it_scenario["triggers"]
        )
    )


   

    print("\n")
    print("#" * 70)
    print("# 2. BANK FRAUD VISHING SCENARIO")
    print("#" * 70)

    print(
        generate_vishing_script(
            bank_scenario["company"],
            bank_scenario["role"],
            bank_scenario["pretext"],
            bank_scenario["name"]
        )
    )

    print(
        analyze_triggers(
            bank_scenario["name"],
            bank_scenario["triggers"]
        )
    )


    

    print("\n")
    print("#" * 70)
    print("# 3. GOVERNMENT SMISHING SCENARIO")
    print("#" * 70)

    print(
        generate_smishing_script(
            government_scenario["company"],
            government_scenario["pretext"],
            government_scenario["link"]
        )
    )

    print(
        analyze_triggers(
            government_scenario["name"],
            government_scenario["triggers"]
        )
    )

   
```
