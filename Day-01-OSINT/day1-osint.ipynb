```python
import whois
import socket
import requests
```

```python
def get_whois_info(domain):
    try:
        w = whois.whois(domain)
        return {
            "registrar": w.registrar,
            "creation_date": str(w.creation_date),
            "expiration_date": str(w.expiration_date)
        }
    except Exception as e:
        return {"error": str(e)}
```

```python
def get_dns_info(domain):
    try:
        ip = socket.gethostbyname(domain)
        return {"ip": ip}
    except Exception as e:
        return {"error": str(e)}
```

```python
def get_geolocation(ip):
    try:
        geo = requests.get(f"http://ip-api.com/json/{ip}", timeout=5).json()
        return {
            "city": geo.get("city"),
            "country": geo.get("country"),
            "isp": geo.get("isp")
        }
    except Exception as e:
        return {"error": str(e)}
```

```python
def osint_scan(domain):
    print(f"\n--- OSINT SCAN: {domain} ---\n")

    whois_data = get_whois_info(domain)
    print("[WHOIS]", whois_data)

    dns_data = get_dns_info(domain)
    print("[DNS]", dns_data)

    ip = dns_data.get("ip")
    geo_data = get_geolocation(ip) if ip else {"error": "no IP found"}
    print("[GEO]", geo_data)

    return {"whois": whois_data, "dns": dns_data, "geo": geo_data}
```

```python
results = osint_scan("example.com")
```
